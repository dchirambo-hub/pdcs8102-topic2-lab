### Task 1.3 - Condition and Function

```python
def get_grade(score):
    if score >=80:
        return "Distinction"
    elif score >70:
        return "Merit"
    elif score >=60:
        return "Pass"
    else:
        return "Fail"
```

```python
print(get_grade(85))
```

```python
print(get_grade(72))
```

```python
print(get_grade(45))
```

## Exercise 3 - Pandas 

```python
import pandas as pd 
data = { 
"StudentID": [101, 102, 103, 104, 105, 106], 
"Name": ["Alice", "Bob", "Charlie", "Diana", "Eve", "Frank"], 
"Age": [25, 30, 35, 28, 32, 27], 
"Score": [85, 72, 90, 68, 95, 78], 
"Department": ["CS", "IT", "CS", "Business", "IT", "CS"] 
} 
df = pd.DataFrame(data)
```

```python
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
print(df.dtypes)
```

## Exercise 4 - Basic visualization (20 minutes)

The purpose is to confirm that you can use the plotting libraries, not to conduct full exploratory data analysis

```python
import matplotlib.pyplot as plt 
plt.bar(df["Name"], df["Score"]) 
plt.title("Student Scores") 
plt.xlabel("Student") 
plt.ylabel("Score") 
plt.xticks(rotation=45) 
plt.tight_layout() 
plt.show()
```

```python
import seaborn as sns 
sns.scatterplot(data=df, x="Age", y="Score") 
plt.title("Age and Score") 
plt.show()
```
